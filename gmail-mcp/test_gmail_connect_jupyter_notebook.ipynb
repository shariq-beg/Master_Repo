```python
from pathlib import Path
from langchain_mcp_adapters.client import MultiServerMCPClient

client = MultiServerMCPClient({
    "gmail": {
        "command": "python",
        "args": [str(Path.cwd() / "server.py")],
        "transport": "stdio",
    }
})
tools = {tool.name: tool for tool in await client.get_tools(server_name="gmail")}

search_result = await tools["search_gmail"].ainvoke({"query": "category:promotions", "max_results": 5})
results = search_result.get("messages", [])
print("Search results:")
print(results)

if results:
    first_id = results[0]["id"]
    print("\nReading first email:\n")
    email_data = await tools["get_gmail_message"].ainvoke({"message_id": first_id})
    print(email_data)
else:
    print("No matching emails found.")
```
